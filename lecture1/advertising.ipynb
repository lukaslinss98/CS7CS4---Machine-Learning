```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from prompt_toolkit.key_binding.bindings.mouse import xterm_sgr_mouse_events
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

```

```python
df = pd.read_csv('Advertising.csv')
df.head()
```

```python
ax = plt.figure().add_subplot(projection='3d')
ax.scatter(xs=df['TV'], ys=df['radio'], zs=df['sales'])
ax.set_xlabel('TV')
ax.set_ylabel('Radio')
ax.view_init(elev=20., azim=-10, roll=0)
plt.show()
```

```python
df.describe()
```

```python
tv_sales = df[['TV', 'sales']]

xtrain = np.array(tv_sales[['TV']]).reshape(-1,1)
ytrain = np.array(tv_sales['sales']).reshape(-1, 1)

xtest = np.arange(0, 300, 1)
```

```python

model = LinearRegression().fit(xtrain, ytrain)
print(model.intercept_, model.coef_)
```

```python
ytest = model.predict(xtest.reshape(-1, 1))
```

```python
plt.scatter(x=xtrain, y=ytrain, color='red')
plt.plot(xtest, ytest, color='blue', linewidth=3)
plt.xlabel('TV')
plt.ylabel('sales')
plt.show()
```
