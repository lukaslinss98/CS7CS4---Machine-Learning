```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import classification_report
```

```python
Xtrain = np.arange(0,1,0.01) # generate values from 0 to 1 with step size 0.01
Xtrain = Xtrain.reshape(-1, 1) # reshape into 100 x 1 matrix (-1 infers value)
Xtrain
```

```python
ytrain = 10 * Xtrain + np.random.normal(0.0, 1.0, 100).reshape(-1,1) # params mean, std-deviation, size
ytrain
```

```python
model = LinearRegression().fit(Xtrain.reshape(-1,1), ytrain.reshape(-1,1))
print(model.intercept_, model.coef_)
ytest = np.arange(0,1,0.01).reshape(-1,1)
ypred = model.predict(ytest)
```

```python
plt.rc('font', size=18)
plt.scatter(Xtrain, ytrain, color='black')
plt.plot(Xtrain, ypred, color='blue', linewidth=3)
plt.xlabel('input x')
plt.ylabel('output y')
plt.legend(['predictions','training data'])
plt.show()
```
