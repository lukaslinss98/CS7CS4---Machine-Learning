```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
```

```python
file = open('../week9_assignment/input_childSpeech_testSet.txt').read()
vocab = sorted(list(set(file)))
len(vocab)
```

```python
stoi = {ch: i for i, ch in enumerate(vocab)}
itos = {i: ch for i, ch in enumerate(vocab)}
encode = lambda s: [stoi[c] for c in s]  # encoder: take a string, output a list of integers
decode = lambda l: ''.join([itos[i] for i in l])  # decoder: take a list of integers, output a string

encoded_file = encode(file)
xs = encoded_file[:5]
ys = encoded_file[1:6]
print(xs, ys)
```

```python
one_hot_tensors = F.one_hot(torch.tensor(xs), num_classes=len(vocab)).float()
# plt.imshow(one_hot_tensors)
# print(one_hot_tensors.shape)
one_hot_tensors
```

```python
w = torch.randn((40,40), requires_grad=True)
print(one_hot_tensors.shape, w.shape)
```

```python
for _ in range(100):
    logits = one_hot_tensors @ w
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)

    ys_prob = probs[torch.arange(5), ys]
    loss = -ys_prob.log().mean()
    print(loss.item())

    w.grad = None
    loss.backward()
    w.data += -0.2 * w.grad
```
